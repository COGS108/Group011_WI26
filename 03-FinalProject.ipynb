## Abstract

This project is an investigation as to whether seasonal patterns in popular music can be detected through Spotify audio features of Billboard Hot 100 songs, with years ranging from 2020 to 2023. This project is motivated by prior work that links seasonal mood variation to music preferece, as shown by Pettijohn et al. (2010), who found that people prefer more energetic and upbeat music during the spring and summer seasons. In this project, we ask whether valence, every, and tempo differ measurably across songs charting in the four different seasons across each year, namely, Winter, Spring, Summer, and Fall. Our hypothesis was that summer songs would score higher in valence and energy while winter songs would trend lower and slower tempo. 

We attempted to answer this question by combining Billboard Hot 100 weekly chart data with Spotify audio features sourced from two distinct Kaggle datasets, in total covering 2020-2023, matching songs on normalized titles and primary artist names. After cleaning and merging, our working dataset contained almost 7000 matched chart entries across the four seasons. We conducted exploratory data analysis through summary statistics, distribution histograms, seasonal violin plots, and a grouped bar chart of median audio features per season. 

Our results, at best, can only partially support the hypothesis, and, in some cases, it directly contradicts it. Energy increased across seasons from Winter to Fall, but valence was highest in Winter and lowest in Spring, which is opposite of what the hypothesis predicted. Across all three features, the violin plots revealed some overlapping distributions between seasons, suggesting that while directional trends may exist, seasonal differences in the audio features of charting songs are at best modest rather than dramatic. These findings, combined with significant data limitations, suggest that seasonal mood effects on music, particularly mainstream charting songs, are may be real, but are subtle.

## Discusison and Conclusion

Our analysis of 6741 matched Billboard Hot 100 chart entries from 2020 to 2023 revealed marginal but directionally interesting seasonal differences in audio features. Valence, an indicator of musical positivity, was highest in Winter and lowest in Spring, with Summer and Fall falling closely in between. Energy was the only measure that showed a more consistent upward trend, rising from 0.59 in the Winter to 0.62 in the Spring, then to 0.64 in the Summer, and finally peaking at 0.67 in the Fall. Tempo is difficult to fully analyze from the chart alone since the compression scale could be a bit misleading, so we will be basing the results off of the raw data. The overall median tempo across all seasons was approximately 120 BPM with a standard deviation of 30.9, suggesting a considerable within-season variability regardless of which season we measure. Critically, the violin plots for all three features showed that songs from different seasons have very similar ranges of values, meaning that while small seasonal trends exist, any individual song can hove high or low measurements for any of the three measurements, regardless of what time of year it charted. 

These findings sit in conflict with our prior research that motivated this project. Pettijohn et al (2010) found that season context shifts music preferences toward energetic and upbeat music in spring and summer, and toward more subdued music in fall and winter. Our energy results partially align with this, where it is lowest in Winter, however, the trend peaks in Fall rather than Summer, which does not match the prior framework built in the background section. Most critically, valence moves in the opposite direction from the hypthesis prediction. So, it was Winter, not Summer, that produced the highest median valence in our dataset. 

This contradiction is not necessarily fatal to the broader idea that our project considered. Winthorst et al (2020), gave a strong counterpoint, finding that season-to-season mood differences in a large gernal population were small and largely individual personality traits-driven rather than seasons themselves. Our results seem to be consistent with this view, that the seasonal differences we ovserve are small, and the distributions overlap substantially. It is very much possible that both findings can be true to some extent simultaneously, where seasonal mood preferences exist at the individual level as Pettijohn et al. describes, but that these preferences are too diverse across the population to produce a clean seasonal signal we could measure in the aggregate data. These thoughts will simply be left here, and based on the observed data, it is possible for others to draw their own conclsions for further testing. 

Additionally, it is also worth considereing a structural explanation for our odd Winter valence findings. The Billboard Hot 100 in December and early January is reliably dominated by holiday/Christmas songs that rechart every year. These songs are highly scored on the valence metric by design, which likely inflates the Winter measurement and may not reflect broader listener mood preferences. This is a confound that future work should account for by filtering all holiday songs seperately. 

Taken together, the results suggest that the relationship between season and mood is more nuanced than a simple "summer is more livelier, louder, and happier" narrative. The chart reflects not just listender mood but also industry release patterns, algorithmic manipulation from social media platforms, and holiday music. Any attempt to use charting data as a proxy for collective emotional state must come to terms and somehow filter these forces that each hold their own competing interests. That being said, some existing trends in the data my still be of value or interesting in their own right, which some might use for things like figuring out when to release their songs, etc. Hence, these trends could be useful and interesting with more complete data. 

Other key limitations constrain how broadly these conclusions can be applied. The most significant is data coverage, or lack thereof. The one thing that set this project back the most, and that impacted how much data we could gather, is the Spotify deprecation of the audio features API endpoint in November of 2024. This would rate limit us when we attempted to gather the bulk of the data, making it so that we could only gather data of hundreds of songs per day until we were forced to wait another 24 hours to continue data collection. This, with severe deadline limitations, made it so that we were unable to retrive all of the data we wanted, instead relying on pre-existing Kaggle datasets. Only 21.5% of Billboard chart rows were matched successfully to audio features, meaning that 28.5% of the data had to be discarded. This is, of course, terrible and weakens our authority of making any definitive claims from this analysis. The matched subset very well may not be representative of the full charting population, and the seasonal biases that were probably present in the Kaggle datasets carried over directly into our seasonal comaprisons. 

The seasonal imbalanced in matched rows furthered this concern. Winter had over 2300 matched entries, compared to only just over 1000 for Fall. This is over a 2x disparity, so, Fall conclusions are based on a substantially smaller and less representative sample, and the higher median energy obsrved in Fall should be approached with scrutiny. Additionally, a further limitation comes directly with our forms of measurements as well. Valence and energy are also algorithmic proxies, not direct measurements from human emotional experience. They capture acoustic properties of audio and will not always reflect how listeners experience a song's "intended" or "designed" mood.  

If this work were to continue past this course, several changes would be made to strengthen the analysis. First, we would take more time to obtain audio features, collecting small amounts per day for a while. Second, the most popular holiday songs should be identified and either removed or analyzed seperately to avoid confounding in our measurements. Third, more formal statistical testing would determine whether the observed seasonal differences are statistically significant or within range of chance variation.
